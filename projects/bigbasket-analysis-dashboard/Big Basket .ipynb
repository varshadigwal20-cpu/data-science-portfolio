# Checking Outliers

```python
Q1 = df["sale_price"].quantile(0.25)
Q3 = df["sale_price"].quantile(0.75)
print("Q1:", Q1)
print("Q3:", Q3)
```

```python
IQR = Q3 - Q1
print(IQR)
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
lower_bound
```

```python
upper_bound
```

```python
df[(df["sale_price"] < lower_bound) | 
   (df["sale_price"] > upper_bound)].shape
```

```python
df[df["sale_price"] > 755].head(5)
```

```python
plt.figure(figsize=(10,6))
plt.hist(df["sale_price"], bins=50,edgecolor='black')
plt.axvline(x=755, color='red', linestyle='--', linewidth=2, label='Upper Boundary (755)')
plt.title("Sale Price Distribution with IQR Upper Boundary", fontsize=16, fontweight='bold')
plt.xlabel("Sale Price", fontsize=12)
plt.ylabel("Frequency", fontsize=12)
plt.legend()
plt.show()
```

```python
top_categories = df["category"].value_counts().head(10)
plt.figure(figsize=(10,6))
top_categories.plot(kind="bar",color="slateblue",edgecolor="black")
plt.title("Top 10 Categories by Product Count",fontsize=16, fontweight='bold')
plt.xlabel("Category",fontsize=14)
plt.ylabel("Number of Products",fontsize=12)
plt.xticks(rotation=60, fontsize=10)
plt.yticks(fontsize=10)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["rating"], bins=20, kde=True)
plt.title("Distribution of Product Ratings",fontsize=16, fontweight='bold')
plt.xlabel("Rating",fontsize=12)
plt.ylabel("Frequency",fontsize=12)
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.title("Market Price vs Sale Price",fontsize=16, fontweight='bold')
plt.xlabel("Market Price",fontsize=12)
plt.ylabel("Sale Price",fontsize=12)
plt.scatter(df["market_price"], df["sale_price"], alpha=0.5)
max_val = max(df["market_price"].max(), df["sale_price"].max())
plt.plot([0, max_val], [0, max_val])
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["discount_percentage"], bins=20)
plt.title("Distribution of Discount Percentage",fontsize=16, fontweight='bold')
plt.xlabel("Discount Percentage",fontsize=12)
plt.ylabel("Frequency",fontsize=12)
plt.show()
```

```python
avg_discount = df.groupby("category")["discount_percentage"].mean().sort_values(ascending=False).head(10)
plt.figure(figsize=(10,6))
avg_discount.plot(kind="barh",color = "teal",edgecolor ="black")
plt.title("Top 10 Categories by Average Discount",fontsize=16, fontweight='bold')
plt.xlabel("Average Discount (%)",fontsize=12)
plt.show()
```
